```python
gastos = [15.0, 57.0, 03.25, 11.75, 2.0, 52.50, 97.75, 72.25]
essencial = [True, False, False, True, True, True, False, True]

gastos_essencial = 0
gastos_nao_essencial = 0

for i in range(len(gastos)):
  if essencial[i]:
    gastos_essencial += gastos[i]
  else:
    gastos_nao_essencial += gastos[1]

print("gastos essencial:", gastos_essencial )
print("gastos não essencial:", gastos_nao_essencial )
```

```python
produtos = ['avela', 'sabão em pó', 'manteiga', 'cebola']
qtd = [1, 1, 1, 1]
menu ="""\

Escolha uma opção:

(1) Alterar quantidade
(2) Sair

"""
opcao = int(input(menu))

while opcao != 2:
  if opcao == 1:
      prod = input("\n Qual Produto ? ")
      if prod in produtos:
          pos = produtos.index(prod)
          quant = qtd[pos]
          print(prod,"tem", quant, "unidade")
          nova_quant = int(input("\n Nova quantidade ? "))
          if nova_quant == 0 :
            produtos.pop(pos)
          else:
              qtd[pos] = nova_quant
      else:
        print("OPS, A lista não contem ")
  else:
      print(opcao,"não é uma opção válida !")
  opcao = int(input(menu))

  for i in range(len(produtos)):
      prod = produtos[i]
      quant = qtd[i]
      print("\n", prod,"tem", quant, "unidade: ")


```
