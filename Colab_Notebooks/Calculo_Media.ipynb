```python
# Lista de nomes abreviados dos meses, usada para montar a data formatada
MESES = ['jan', 'fev', 'mar', 'abr', 'mai', 'jun',
         'jul', 'ago', 'set', 'out', 'nov', 'dez']


def formatar_data(data):
    """
    Recebe uma tupla (ano, mes, dia) e retorna uma string no formato dd/mes/aaaa,
    onde 'mes' é substituído pelas três primeiras letras do nome do mês.
    Ex.: formatar_data((2020, 1, 3)) -> "03/jan/2020"
    """
    ano, mes, dia = data  # desempacota a tupla em três variáveis
    # :02d garante que o dia sempre tenha 2 dígitos (ex.: 3 vira "03")
    # mes - 1 porque a lista MESES começa no índice 0 (jan = índice 0)
    return f"{dia:02d}/{MESES[mes - 1]}/{ano}"


def formatar_hora(hora):
    """
    Recebe uma tupla (hora, minuto) e retorna uma string no formato hh:mm (24h).
    Ex.: formatar_hora((16, 30)) -> "16:30"
    """
    h, m = hora  # desempacota a tupla em hora e minuto
    # :02d garante 2 dígitos tanto para hora quanto para minuto
    return f"{h:02d}:{m:02d}"


def imprimir_eventos(eventos, de_data=(1, 1, 1), ate_data=(9999, 12, 31)):
    """
    Imprime os eventos da agenda cuja data esteja entre de_data e ate_data
    (incluindo os limites).

    Parâmetros:
        eventos  -> lista de tuplas no formato ((ano, mes, dia), (hora, minuto), descricao)
        de_data  -> tupla (ano, mes, dia) que representa a data inicial (padrão: bem antiga)
        ate_data -> tupla (ano, mes, dia) que representa a data final (padrão: bem futura)
    """
    for data, hora, descricao in eventos:
        # Como tuplas em Python são comparadas elemento a elemento
        # (primeiro o ano, depois o mês, depois o dia), a comparação abaixo
        # já verifica corretamente se a data do evento está no intervalo,
        # sem precisar comparar ano, mês e dia manualmente.
        if de_data <= data <= ate_data:
            # Formata data e hora usando as funções auxiliares
            # e imprime no formato: dd/mes/aaaa - hh:mm: descricao
            print(f"{formatar_data(data)} - {formatar_hora(hora)}: {descricao}")


# ------------------------------
# Exemplo de uso
# ------------------------------

agenda = [
    ((2020, 1, 13), (11, 50), 'Renovar identidade'),
    ((2020, 1, 15), (16, 30), 'Fazer compras'),
    ((2020, 1, 25), (8, 45), 'Autenticar documentos'),
    ((2020, 2, 29), (14, 15), 'Prestar concurso'),
    ((2020, 3, 15), (17, 50), 'Buscar bolo pro aniversário da vovó'),
    ((2020, 3, 17), (13, 20), 'Consulta de revisão com dentista'),
]

print("Eventos a partir de 20/jan/2020:")
imprimir_eventos(agenda, (2020, 1, 20))

print("\nEventos até 15/mar/2020:")
imprimir_eventos(agenda, ate_data=(2020, 3, 15))

print("\nTodos os eventos (sem restrição de data):")
imprimir_eventos(agenda)
```
