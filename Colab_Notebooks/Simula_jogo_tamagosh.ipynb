1. construir uma classe para abstrair um animal qualquer.
Nessa abstração:
 inclua os atributos nome, espécie e fome
Por exemplo, nome e espécie podem ser passados como parâmetro no construtor dessa classe;

já a fome deve ser inicializada com valor 0
obs: pois ela é o atributo indicador numérico que armazenará o quanto de fome o animal tem em um dado momento

0 (zero) fome significa que ele está satisfeito,
1, significa que ele tem um pouquinho de fome;
2, um pouco mais; e assim por diante.

2.  Como comportamento:

construir o método __str__() para retornar uma string, representando o estado atual de um dado animal.

método andar(), que simula um passeio com o animal. Ao andar, o valor da fome do animal deve aumentar em 1.

construir um método comer(), que implementa a simulação do ato do animal de se alimentar.

Esse método recebe um valor numérico, como parâmetro, que é o número de unidades de comida oferecidas ao animal.

Com esse valor: você vai diminuir o atributo fome do animal.

Caso o valor de fome do animal seja menor do que o valor de comida oferecido, você deve atualizar o valor de fome para 0

informar ao usuário que:  o animal comeu até ficar saciado e deixou o resto da comida no prato

3. fora da classe
instancie um objeto Animal: com o nome e espécie que você quiser.

4. criar Laço de repetição:
interagir com o usuário:
Ademais, deve haver um menu com 4 opções:

41. para alimentar o animal;
42. para andar com ele;
43. para mostrar o estado atual do animal;
44. para finalizar a execução.

5. O usuário vai digitar a opção que quer, e você vai fazer com que o animal se comporte adequadamente.

6. Na opção 1, você vai apresentar uma nova questão ao usuário para que ele informe o quanto de comida é desejada para dar ao animal.

7. Assim: você vai tratar a possível exceção gerada
informando ao usuário que ele deu comida demais

caso seja a situação: Ao final de cada interação

você vai apresentar o novo estado do animal e,
depois,
apresentar o menu novamente ao usuário

até que ele aperte 3 e finalize a execução do programa.








```python

class ComidaDemaisError(Exception): #Herdando exception (a classe-mãe de todos os erros em Python)
  pass # para que a classe Exception não precise de nenhum tipo de comportamento
class Animal: # classe animal passando por parâmetros nome e especie e inicializando fome com zero.
  def __init__(self, nome, especie):
    self.nome = nome
    self.especie = especie
    self.fome = 0

#Comportamento
  #metodo que recebe string
  #ele é chamado
  # automaticamente quando você faz print(objeto) ou str(objeto).
  # Precisa retornar (return) uma string.
  def __str__(self):# -> str:
    return f'{self.nome} é um(a) {self.especie} com nível de fome {self.fome}'
  #metodo andar
  # Cada chamada de andar() representa UM passeio. Um passeio aumenta
  # a fome em 1 unidade. self.fome += 1 é o mesmo que self.fome = self.fome + 1
  def andar(self):
    self.fome += 1
  # comer() recebe de fora a quantidade de comida oferecida
  def comer(self, quantidade):

    if  quantidade > self.fome:
        self.fome = 0 # zera a fome (ficou saciado)
        raise ComidaDemaisError(
            f'{self.nome} comeu até ficar saciado e deixou o resto da comida'
        )
    else:
        self.fome -= quantidade


meu_animal = Animal('Rock', 'cachorro')

while True:
  print('\n ====MENU====')
  print('Digite [1] para alimentar o Animal: ')
  print('Digite [2] para andar com ele: ')
  #print('Digite [3] para mostrar o estado atual do Animal: ')
  print('Digite [3] para Sair: ')
# come
#andar

  opcao = input('Escolha uma opção: \n ')
  if opcao == '1':
    try:
      quantidade = float(input('Quantas unidades de comida deseja oferecer ? '))
      meu_animal.comer(quantidade)
    except ComidaDemaisError as erro:
      print(f'Voce deu comida demais! {erro}')
    except ValueError:
      print('Digite um número Válido! ')

  elif opcao == '2':
    meu_animal.andar()
    #print('Vocês foram passear! ')

  elif opcao == '3':
    #print(meu_animal)
    print('Sair')
    break
  print(meu_animal)
  #elif opcao == '4':
  #  print('Sair')
  #  break





```
