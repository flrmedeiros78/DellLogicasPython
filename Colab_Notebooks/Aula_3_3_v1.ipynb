```python

#miau para ser chamado do google drive para executar po
import datetime as dt

class Gato:
  """Um bicho"""
  def __init__(self, nome, sexo, peso, nasc):
    """Nome (str), sexo ('m'ou'f'), peso (float),
    nasc (datetime.date)."""
    self.nome = nome
    self.sexo = sexo
    self.peso = peso
    self.nasc = nasc
  def __str__(self):
    return '{}, sexo{}, {}kg, nasceu em {}'.format(self.nome, self.sexo, self.peso, self.nasc)
  def idade(self):
    """ Idade do bicho, em anos (ponto-flutuante)."""
    return (dt.today() - self.nasc).days / 365
```
